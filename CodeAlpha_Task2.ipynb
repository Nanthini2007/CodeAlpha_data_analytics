```python
# 2.TOP 10 CHANNELS BY SUBSCRIBERS
top_10=df.sort_values(by='subscribers',ascending=False)
top_10[["Youtuber", "subscribers"]].head(10)
```

```python
#STEP10:VISUALIZATION
```

```python
#1.Top Categories
plt.figure(figsize=(10,5))
sns.countplot(y="category",data=df, order=df["category"].value_counts().index)
plt.title("Top YouTube Categories")
plt.show()
```

```python
#2.Subscribers vs Video Views
plt.figure(figsize=(8,5))
sns.scatterplot(x="subscribers", y="video views", data=df)
plt.title("Subscribers vs Video Views")
plt.show()
```

```python
#3.Channel Creation Trend
year_data = df["created_year"].value_counts().sort_index()
plt.figure(figsize=(10,5))
plt.plot(year_data.index,year_data.values)
plt.title("YouTube Channel Creation Over Years")
plt.xlabel("Year")
plt.ylabel("Number of Channels")
plt.show()
```

```python
#Step13: Detect Outliers
plt.figure(figsize=(10,5))
sns.boxplot(x=df["subscribers"])
plt.title("Subscribers Outliers")
plt.show()
```
